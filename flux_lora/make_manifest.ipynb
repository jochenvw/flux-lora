```python
import os
import json

json_lines = []
files = [f for f in os.listdir("../work/dataset/new") if f.endswith(".jpg")]
for file in files:
    filename = os.path.splitext(file)[0]
    name_only = os.path.basename(filename)

    txt_path = f"../work/dataset/new/{filename}.txt"
    with open(txt_path, "r", encoding="utf-8") as f:
        caption = f.read().strip()

    caption += " Style = #Ah2025"

    # construct JSON line with properties 'image', 'conditioning_image' and 'text'
    json_line = {
        "name": name_only,
        "image": f"../work/dataset/new/{filename}.jpg",
        "conditioning_image": f"../work/dataset/new/{filename}.jpg",
        "text": caption
    }

    json_lines.append(json_line)

with open("./train.jsonl", "w", encoding="utf-8") as f:
    for line in json_lines:
        f.write(json.dumps(line, ensure_ascii=False) + "\n")
```

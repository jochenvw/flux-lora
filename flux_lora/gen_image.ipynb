```python
from diffusers import FluxControlPipeline
import torch
from PIL import Image

# Load the pipeline with your trained LoRA
pipe = FluxControlPipeline.from_pretrained(
    "black-forest-labs/FLUX.1-dev",
    torch_dtype=torch.float16
).to("cuda")
pipe.load_lora_weights("output/lora-ah2025")

# Load your input image
init_image = Image.open("../work/converted/old/courgettebroodjes.jpg").convert("RGB")
init_image = init_image.resize((512, 512))  # match training resolution

# Define style prompt
prompt = "in style #Ah2025 - courgettebroodjes - top-down composition, simple minimal background, soft natural lighting, vibrant color palette, smooth textures, cheerful and inviting mood, food photography"

# Generate styled image
image = pipe(
    prompt=prompt,
    control_image=init_image,   # <--- the input image
    num_inference_steps=30,
    guidance_scale=7.5,
    height=512,
    width=512
).images[0]

image.save("courgettebroodjes.png")
```
